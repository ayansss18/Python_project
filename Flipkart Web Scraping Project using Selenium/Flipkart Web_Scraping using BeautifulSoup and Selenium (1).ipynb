# Using loop to automate

```python
url='https://www.flipkart.com/search?q=laptops+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page=1'
path='msedgedriver.exe'
s=Service(path)
driver=webdriver.Edge(service=s)
driver.get(url)
time.sleep(5)
data=driver.page_source
# driver.quit()
soup=BeautifulSoup(data,'lxml')

i=0
while i<=10:
    nxt_pg=soup.find('a',class_='_9QVEpD')
    cp_nxt_pg='https://www.flipkart.com'+nxt_pg['href']
    print(cp_nxt_pg)
    
    url=cp_nxt_pg
    driver.get(url)
    # time.sleep(0)
    data=driver.page_source
    
    soup=BeautifulSoup(data,'lxml')
    i+=1

driver.quit()
```

# Get Specific Links for continuous page

```python
path='msedgedriver.exe'
s=Service(path)
driver=webdriver.Edge(service=s)

for i in range(2 ,10):

    url='https://www.flipkart.com/search?q=laptops+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page='+str(i)
    
    driver.get(url)
    time.sleep(1)
    page=driver.page_source
    soup=BeautifulSoup(page,'lxml')
    nxt_pg=soup.find('a',class_='_9QVEpD')['href']
    cmp_nxt_pg="https://www.flipkart.com"+nxt_pg
    print(cmp_nxt_pg)
    
driver.quit()
```

# Extracting Data

```python
url='https://www.flipkart.com/search?q=laptops+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page='+str(1)
path='msedgedriver.exe'
s=Service(path)
driver=webdriver.Edge(service=s)
driver.get(url)
time.sleep(1)
page=driver.page_source
driver.quit()
```

```python
soup=BeautifulSoup(page,'lxml')
print(soup.prettify())
```

```python
Laptop_names=[]
Laptop_price=[]
Laptop_rating=[]

names=soup.find_all('div',class_='tUxRFH')
# print(names)
for name in names:
    lap_names=name.find('div',class_='KzDlHZ')
    lap_names_txt=lap_names.text
    Laptop_names.append(lap_names_txt)
    
    price=name.find('div',class_='Nx9bqj _4b5DiR')
    price_txt=price.text
    Laptop_price.append(price_txt)
    
    rating=name.find('div',class_='XQDdHH')
    rating_txt=rating.text if rating else "N/A"
    Laptop_rating.append(rating_txt)
                     
    print(lap_names_txt,price_txt,rating_txt)
```

```python
import pandas as pd
```

```python
data=pd.DataFrame({
    'Laptop_names':Laptop_names,
    'Laptop_price':Laptop_price,
    'Laptop_rating':Laptop_rating
})
```

# Final Project - Automating to get upto 5 pages data and save to CSV File

```python
Laptop_names=[]
Laptop_price=[]
Laptop_rating=[]

path='msedgedriver.exe'
s=Service(path)
driver=webdriver.Edge(service=s)

for i in range(1,6):
    url='https://www.flipkart.com/search?q=laptops+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page='+str(i)
    
    driver.get(url)
    time.sleep(1)
    page=driver.page_source
    soup=BeautifulSoup(page,'lxml')

    
    names=soup.find_all('div',class_='tUxRFH')
    # print(names)
    for name in names:
        lap_names=name.find('div',class_='KzDlHZ')
        lap_names_txt=lap_names.text
        Laptop_names.append(lap_names_txt)
        
        price=name.find('div',class_='Nx9bqj _4b5DiR')
        price_txt=price.text
        Laptop_price.append(price_txt)
        
        rating=name.find('div',class_='XQDdHH')
        rating_txt=rating.text if rating else "N/A"
        Laptop_rating.append(rating_txt)
                         

driver.quit()

data=pd.DataFrame({
    'Laptop_names':Laptop_names,
    'Laptop_price':Laptop_price,
    'Laptop_rating':Laptop_rating
})
print(data)
```

```python
data.to_csv("Flipkart_webScraping.csv")
```
