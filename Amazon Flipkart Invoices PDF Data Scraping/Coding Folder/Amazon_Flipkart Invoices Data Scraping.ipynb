# Creating a common function to extract data

```python
def extract_data1(path):
    data=[]
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            text=page.extract_text()
            lines=text.split('\n')
            for line in lines:
                data.append(line.strip())
            
    return data
```

# Amazon Invoices Scraping

```python
def amazon_extracted_invoice_text(lines):
    parsed_data={
        'Invoice Number':None,
        'PAN No':None,
        'Invoice Date':None,
        'Place of delivery':None,
        'Payment Transaction ID':None,
        'Mode of Payment':None,
        'Date & Time':None,
        "GST Registration No":None
    }

    for line in lines:
        if 'Invoice Number' in line:
            parsed_data['Invoice Number']=line.split(":")[1].strip().split()[0]
        elif 'PAN No' in line:
            parsed_data["PAN No"] = line.split(":")[1].strip().split()[0]
        elif 'Invoice Date' in line:
            parsed_data["Invoice Date"] = line.split(":")[1].strip()
        elif 'Place of delivery' in line:
            parsed_data["Place of delivery"] = line.split(":")[1].strip()
        elif 'Payment Transaction ID' in line:
            parsed_data["Payment Transaction ID"] = line.split(":")[1].strip()
        elif 'Date & Time' in line:
            parsed_data["Date & Time"] = line.split(":")[1].strip()
        elif 'Mode of Payment' in line:
            parsed_data["Mode of Payment"] = line.split(":")[-1].strip()
        elif 'GST Registration No' in line:
            parsed_data["GST Registration No"] = line.split(":")[1].strip().split()[0]
    return parsed_data
```

### Extracting data from all Amazon invoices files

```python
Amazon_files=['Amazon_invoice_1.pdf','Amazon_invoice_2.pdf','Amazon_invoice_3.pdf','Amazon_invoice_4.pdf']
amazoninvoices=[]

for file in Amazon_files:
    raw_data=extract_data1(file)
    final_data=amazon_extracted_invoice_text(raw_data)
    amazoninvoices.append(final_data)
```

### Saving the scraped data to an excel file

```python
df=pd.DataFrame(amazoninvoices)
df.to_excel('Amazon_invoice_data.xlsx',index=False)
```

# Flipkart Invoices Scraping

```python
def flipkart_extracted_invoice_text(lines):
    parsed_data={
        'Invoice Number':None,
        'PAN':None,
        'GSTIN':None,
        'Invoice Date':None,
        'CIN':None,
        'Grand Total':None,
        "Sold By":None
    }

    for line in lines:
        if 'Invoice Number' in line: 
            parsed_data['Invoice Number']=line.split("#")[1].strip()
        elif 'PAN' in line:
            parsed_data["PAN"] = line.split(":")[1].strip().split()[0]
        elif 'GSTIN' in line:
            parsed_data["GSTIN"] = line.split("-")[1].strip()
        elif 'Invoice Date' in line:
            parsed_data["Invoice Date"] = line.split(":")[1].strip()
        elif 'CIN' in line:
            parsed_data["CIN"] = line.split(":")[1].strip()
        elif 'Grand Total' in line:
            parsed_data["Grand Total"] = line.split()[3].strip()
        elif 'Sold By' in line:
            parsed_data["Sold By"] = line.split(":")[1].strip()
    return parsed_data
```

### Extracting data from all Flipkart invoices files

```python
Flipkart_files=['Flipkart_invoice_1.pdf','Flipkart_invoice_2.pdf','Flipkart_invoice_3.pdf','Flipkart_invoice_4.pdf']
flipkartinvoices=[]

for file in Flipkart_files:
    raw_data=extract_data1(file)
    final_data=flipkart_extracted_invoice_text(raw_data)
    flipkartinvoices.append(final_data)
    
    
```

#### Saving the scraped data to an excel file

```python
df=pd.DataFrame(flipkartinvoices)
df.to_excel("Flipkart_invoice_data.xlsx",index=False)
```
