# Working on Real Website - Internshala

```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
r=requests.get("https://internshala.com/internships/data-science-internship/")
print(r)
```

```python
soup=BeautifulSoup(r.content,'lxml')
print(soup.prettify())
```

```python
os.makedirs('Job_details',exist_ok=True)
```

```python

jobs=soup.find_all('div',class_='internship_meta duration_meta')

for index, job in enumerate(jobs):
    job_title=job.find("a")
    final_job=job_title.text
    company_name=job.find('p',class_='company-name')
    company=company_name.text.strip()
    stipend=job.find('span',class_='stipend')
    stipend=stipend.text
    job_posted=job.find('div',class_='color-labels')
    job_posted=job_posted.span.text

    
    with open(f'Job_details/Job{index}.txt','w',encoding='utf-8') as f:
        f.write(f"Company name - {company} \n")
        f.write(f'Job Position Available - {final_job}\n')
        f.write(f'Salary range - {stipend}\n')
        f.write(f'Job posted - {job_posted}\n')

    
print('File Saved Successfully...')
```
